```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/processed/immigrant_data_cleaned.csv")
```

## RENDA MÉDIA POR INGLÊS

```python
import matplotlib.pyplot as plt

result = df.groupby("english_level")["annual_income"].mean()

result.plot(kind="bar")

plt.title("Average Income by English Level")
plt.xlabel("English Level")
plt.ylabel("Average Income")
plt.show()
```

## ENTRADA vs RENDA

```python
df.groupby("entry_type")["annual_income"].mean().plot(kind="bar")
plt.title("Income by Entry Type")
plt.show()
```

## PROFISSÃO ANTERIOR vs ATUAL

```python
pd.crosstab(df["previous_profession"], df["current_profession"])
```

## HORAS vs ESTRESSE

```python
plt.scatter(df["hours_per_week"], df["stress_level"])
plt.xlabel("Hours per Week")
plt.ylabel("Stress Level")
plt.show()
```

## CORRELAÇÃO

```python
df.corr(numeric_only=True)
```
